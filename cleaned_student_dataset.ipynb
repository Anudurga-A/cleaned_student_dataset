```python
import pandas as pd
import numpy as np

print("Libraries imported successfully!")
```

```python
data = {
    "Name": ["Anu", "Priya", "Kaviya", "Rahul", "Anu", "Vijay"],
    "Age": [21, 22, np.nan, 23, 21, 24],
    "Course": ["MCA", "MCA", "MBA", "MCA", "MCA", None],
    "Marks": [85, 90, 88, np.nan, 85, 78]
}

df = pd.DataFrame(data)

print("Original Dataset:")
display(df)
```

```python
print("Dataset Information:")
df.info()

print("\nDataset Shape:")
print(df.shape)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())
```

```python
# Fill missing age with the average age
df["Age"] = df["Age"].fillna(df["Age"].mean())

# Fill missing course with the most frequent course
df["Course"] = df["Course"].fillna(df["Course"].mode()[0])

# Fill missing marks with the average marks
df["Marks"] = df["Marks"].fillna(df["Marks"].mean())

print("Missing Values Handled:")
display(df)
```

```python
df = df.drop_duplicates()

print("Dataset After Removing Duplicates:")
display(df)

print("Remaining Duplicate Rows:", df.duplicated().sum())
```

```python
df["Age"] = df["Age"].round().astype(int)
df["Marks"] = df["Marks"].round(2)

print("Data Types After Correction:")
print(df.dtypes)
```

```python
print("Missing Values After Cleaning:")
print(df.isnull().sum())

print("\nDuplicate Rows After Cleaning:")
print(df.duplicated().sum())

print("\nFinal Cleaned Dataset:")
display(df)
```

```python
df.to_csv("cleaned_student_dataset.csv", index=False)

print("Cleaned dataset saved successfully!")
```
